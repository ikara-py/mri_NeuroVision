```python
from pathlib import Path
from PIL import Image
import kagglehub
```

```python
path = kagglehub.dataset_download(
    "ahmedsorour1/mri-for-brain-tumor-with-bounding-boxes"
)

DATA_DIR = Path(path)

IMAGE_EXTENSIONS = {".jpeg", ".jpg", ".bmp", ".png"}

images = list(DATA_DIR.rglob("*"))
images = [file for file in images if file.is_file() and file.suffix.lower() in IMAGE_EXTENSIONS]

labels = list(DATA_DIR.rglob("*.txt"))
labels = [file for file in labels if file.is_file()]

print("Images found:", len(images))
print("Label files found:", len(labels))
```

```python
image_stems = {image.stem for image in images}
label_stems = {label.stem for label in labels}

orphan_images = [image for image in images if image.stem not in label_stems]
orphan_labels = [label for label in labels if label.stem not in image_stems]

print("Images without labels:", len(orphan_images))
print("Labels without images:", len(orphan_labels))

# Keep only images that have a matching .txt file
valid_images = [image for image in images if image.stem in label_stems]

print("Images with labels:", len(valid_images))

```

```python
clean_images = []

for image_path in valid_images:
    try:
        # Check whether the image can be opened
        with Image.open(image_path) as img:
            img.verify()

        clean_images.append(image_path)

    except Exception as error:
        print("Corrupted image:", image_path.name)

print("Valid images:", len(clean_images))
print("Corrupted images:", len(valid_images) - len(clean_images))
```
